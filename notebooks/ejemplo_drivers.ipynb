# Ejemplo: leer `drivers` desde MySQL (`f1db`)

Este notebook se conecta al contenedor **MySQL** (`mysql` en la red de Compose),
usa la base **`f1db`** y muestra las **10 primeras filas** de la tabla `drivers`.

Requisito: el conector JDBC debe estar en `spark_jars/` (ver README):
`mysql-connector-j-8.4.0.jar`

```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("f1db-drivers")
    .master("local[2]")
    .config("spark.driver.memory", "3g")
    .config("spark.driver.cores", "2")
    .config("spark.jars", "/opt/spark/jars/mysql-connector-j-8.4.0.jar")
    .getOrCreate()
)

spark
```

```python
jdbc_url = (
    "jdbc:mysql://mysql:3306/f1db"
    "?useSSL=false"
    "&allowPublicKeyRetrieval=true"
    "&serverTimezone=UTC"
)

drivers = (
    spark.read.format("jdbc")
    .option("url", jdbc_url)
    .option("dbtable", "drivers")
    .option("user", "f1user")
    .option("password", "f1password")
    .option("driver", "com.mysql.cj.jdbc.Driver")
    .load()
)

drivers.show(10)
```

```python
races = (
    spark.read.format("jdbc")
    .option("url", jdbc_url)
    .option("dbtable", "races")
    .option("user", "f1user")
    .option("password", "f1password")
    .option("driver", "com.mysql.cj.jdbc.Driver")
    .load()
)

races.show(10)
```
